## 2. Importar librerías y abrir el portafolio

Importar las librerías con sus alias estándar y crear las estructuras donde el grupo guardará su portafolio.

## 18. Descargar los precios del portafolio

Traer en una sola tabla los precios de cierre de todos los activos que propuso el grupo.

```python
# Precios de cierre de los activos que propuso la clase
precios = yf.download(activos, start="2023-01-01", end="2025-01-01", interval="1mo")

print("Dimensiones:", precios.shape)
precios.head()
```

## 19. Graficar la distribución de los precios

Ver con matplotlib cómo se mueve el precio de cada activo y cómo se reparten esos precios en un histograma.

```python
# Dos miradas al mismo dato: la línea en el tiempo y la distribución
fig, ejes = plt.subplots(1, 2, figsize=(13, 4.5))

# Izquierda: evolución del precio de cierre
precios.plot(ax=ejes[0])
ejes[0].set_title("Precio de cierre del portafolio")
ejes[0].set_xlabel("Fecha")
ejes[0].set_ylabel("Precio (USD)")

# Derecha: distribución de los precios (completa el tipo de gráfico)
precios.hist(ax=ejes[1], bins=15, alpha=0.6)
ejes[1].set_title("Distribución de los precios")
ejes[1].set_xlabel("Precio (USD)")
ejes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

## 20. Rendimientos periódicos

Convertir los precios en rendimientos mensuales para poder comparar activos de precios muy distintos.

```python
# Rendimientos mensuales del portafolio
rendimientos = precios.pct_change().dropna()

(rendimientos * 100).round(2).head()
```

## 21. Graficar los rendimientos y su distribución

Dibujar los rendimientos en el tiempo y su histograma para ver de un vistazo qué tan movido es el portafolio.

```python
# Rendimientos en el tiempo y cómo se reparten
fig, ejes = plt.subplots(1, 2, figsize=(13, 4.5))

# Izquierda: rendimiento mes a mes
(rendimientos * 100).plot(ax=ejes[0])
ejes[0].axhline((0), color="black", linewidth=1)   # línea del cero: ganar o perder
ejes[0].set_title("Rendimiento mensual (%)")
ejes[0].set_xlabel("Fecha")
ejes[0].set_ylabel("Rendimiento (%)")

# Derecha: distribución de los rendimientos
(rendimientos * 100).hist(ax=ejes[1], bins=20, alpha=0.6)
ejes[1].set_title("Distribución de los rendimientos")
ejes[1].set_xlabel("Rendimiento (%)")
ejes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

## 22. Riesgo: dispersión y volatilidad

Medir con un mismo criterio qué tan movido es cada activo del portafolio.

```python
# Rendimiento medio y volatilidad por activo
resumen = pd.DataFrame({
    "rendimiento_medio_%": (rendimientos.mean() * 100).round(2),
    "volatilidad_mensual_%": (rendimientos.std() * 100).round(2),
    "volatilidad_anual_%": (rendimientos.std() * np.sqrt(12) * 100).round(2),
})

resumen.sort_values("volatilidad_anual_%", ascending=False)
```

## 23. Correlación y decisión del portafolio

Comprobar con la matriz de correlación si la diversificación por sectores funcionó de verdad.

```python
# ¿Se movieron juntos o se compensaron entre sí?
matriz = rendimientos.corr().round(3)

print("Sectores del portafolio:", sectores)
matriz
```
