## 2. Variables y tipos de datos

Una **variable** es un espacio con nombre donde guardamos un valor.

| Tipo | Python | R | Ejemplo |
|---|---|---|---|
| Texto (string) | `str` | `character` | `"Cuaderno"` |
| Entero | `int` | `numeric` | `5` |
| Decimal | `float` | `numeric` | `2500.50` |
| Lógico | `bool` | `logical` | `True` / `TRUE` |

En **Python** las variables se asignan con `=`.
En **R** normalmente se usa `<-` (aunque `=` también funciona).


#### 🔎 `print()` con varios valores

`print()` puede recibir **varios valores separados por comas**; los muestra todos en la misma línea, separados por un espacio automáticamente. `type(valor)` nos dice de qué tipo es una variable (texto, número entero, booleano...).


````python
# PYTHON
nombre_producto = "Cuaderno cuadriculado 100 hojas"
precio_unitario = 3500        # int
cantidad = 4                  # int
disponible = True             # bool

print(nombre_producto, precio_unitario, cantidad, disponible)
print(type(nombre_producto), type(precio_unitario), type(disponible))
````

En R, `cat()` funciona igual: recibe varios valores separados por coma y los muestra en una sola línea (por eso agregamos `"\n"` al final, para el salto de línea). `class(valor)` es el equivalente de `type()`.


````python
%%R
# R
nombre_producto <- "Cuaderno cuadriculado 100 hojas"
precio_unitario <- 3500
cantidad <- 4
disponible <- TRUE

cat(nombre_producto, precio_unitario, cantidad, disponible, "\n")
print(class(nombre_producto)); print(class(precio_unitario)); print(class(disponible))
````

#### 🔎 Formateando decimales: `:.0f`

Dentro de una f-string, después de la variable puedes agregar `:` seguido de un **formato**. `:.0f` significa "muéstralo como número decimal (`f` = *float*), pero con **0 cifras decimales**" — así `14000.0` se muestra como `14000`. Muy útil para valores en pesos, donde no queremos ver decimales.


````python
# PYTHON
IVA = 0.19
iva_producto_1 = subtotal_producto_1 * IVA
total_producto_1 = subtotal_producto_1 + iva_producto_1

print(f"Subtotal: ${subtotal_producto_1}")
print(f"IVA (19%): ${iva_producto_1:.0f}")
print(f"Total: ${total_producto_1:.0f}")
````

### 🧾 Aplicado a la papelería
Regla de negocio de la papelería:

- Compras **≥ $50.000** → 10% de descuento
- Compras **≥ $30.000** → 5% de descuento
- Compras menores → sin descuento
- Además, si el cliente es **estudiante**, se suma 2% extra de descuento

Vamos a aplicar esta regla al total del producto 1.


````python
# PYTHON
es_estudiante = True

if total_producto_1 >= 50000:
    descuento = 0.10
elif total_producto_1 >= 30000:
    descuento = 0.05
else:
    descuento = 0.0

if es_estudiante:
    descuento += 0.02

valor_descuento = total_producto_1 * descuento
total_final = total_producto_1 - valor_descuento

print(f"Descuento total: {descuento*100:.0f}%")
print(f"Valor del descuento: ${valor_descuento:.0f}")
print(f"TOTAL A PAGAR: ${total_final:.0f}")
````

# **Nota clase clase 7 liquidacion de la factura**

````python
# Papelería El Lápiz Feliz — liquidación de factura
# S07 · Estructuras condicionales — Python
# Código escrito por el estudiante en clase.

# ── 1 · Los datos de la venta ──
# Datos de la venta
precio = 23000
cantidad = 25
categoria = "papel"
tipo_cliente = "colegio"

subtotal = precio * cantidad           # completa: precio por cantidad
print(subtotal)

# ── 2 · Decisión en escalera: descuento por volumen ──
# Descuento por volumen
if cantidad >= 50:
    pct_volumen = 0.15
elif cantidad >= 20:
    pct_volumen = 0.10    # completa
else:
    pct_volumen = 0     # completa

print(pct_volumen)

# ── 3 · Condición compuesta: descuento por convenio ──
# Descuento por convenio
if tipo_cliente == "colegio"   or     tipo_cliente == "detal":   # une las dos preguntas
    pct_convenio = 0.05
else:
    pct_convenio = 0.00

print(pct_convenio)

# ── 4 · Decisión múltiple: la tarifa de IVA ──
# Tarifa de IVA
if categoria == "libros"          :   # completa la categoría exenta
    pct_iva = 0.0
elif categoria == "revistas":
    pct_iva = 0.05      # completa
else:
    pct_iva = 0.19       # completa

print(pct_iva)

# ── 5 · El programa completo: total a pagar ──
# Liquidación de la factura
dcto_volumen =   subtotal * pct_volumen
dcto_convenio =  (subtotal - dcto_volumen) * pct_convenio
base =   subtotal - dcto_volumen - dcto_convenio
iva = base * pct_iva
total =  base + iva

print("Subtotal:", round(subtotal))
print("Dcto. volumen:", round(dcto_volumen))
print("Dcto. convenio:", round(dcto_convenio))
print("IVA:", round(iva))
print("TOTAL:", round(total))
````

#### 🔎 Alineando texto en columnas: `:<20` y `:>10`

Para que una tabla se vea ordenada, le decimos a Python cuántos caracteres debe ocupar cada valor y hacia qué lado alinearlo:
- `:<20` → alinea el texto a la **izquierda**, dejando espacio hasta completar 20 caracteres.
- `:>10` → alinea a la **derecha**, hasta completar 10 caracteres.

Así, `f"{nombre:<20}"` deja el nombre pegado a la izquierda y rellena con espacios a la derecha, para que la siguiente columna siempre empiece en la misma posición — igual que una tabla de Excel con columnas de ancho fijo.


````python
# PYTHON
nombres    = ["Cuaderno", "Lápiz", "Borrador", "Resma de papel"]
precios    = [3500, 800, 600, 15000]
cantidades = [4, 10, 2, 1]

total_compra = 0
print("DETALLE DE LA COMPRA")
print("-" * 40)
for i in range(len(nombres)):
    subtotal = precios[i] * cantidades[i]
    total_compra += subtotal
    print(f"{nombres[i]:<20} x{cantidades[i]:<3} -> ${subtotal}")

print("-" * 40)
print(f"TOTAL: ${total_compra}")
````

En R, la alineación va dentro del propio código de `sprintf()`: `%-20s` alinea el texto a la **izquierda** en un ancho de 20 caracteres (el `-` es lo que indica "izquierda"; sin él, `%20s` alinearía a la derecha). Y `strrep("-", 40)` es el equivalente en R al truco de Python `"-" * 40`: repite el carácter `"-"` 40 veces para crear la línea separadora.


````python
%%R
# R
nombres    <- c("Cuaderno", "Lápiz", "Borrador", "Resma de papel")
precios    <- c(3500, 800, 600, 15000)
cantidades <- c(4, 10, 2, 1)

total_compra <- 0
cat("DETALLE DE LA COMPRA\n")
cat(strrep("-", 40), "\n")
for (i in 1:length(nombres)) {
  subtotal <- precios[i] * cantidades[i]
  total_compra <- total_compra + subtotal
  cat(sprintf("%-20s x%-3d -> $%d\n", nombres[i], cantidades[i], subtotal))
}
cat(strrep("-", 40), "\n")
cat(sprintf("TOTAL: $%d\n", total_compra))
````

### 🧾 Aplicado a la papelería
Convirtamos en funciones las dos reglas de negocio que ya definimos: **buscar un producto por código** y **calcular el descuento**.


````python
# PYTHON
def buscar_producto(codigo, catalogo):
    for p in catalogo:
        if p["codigo"] == codigo:
            return p
    return None

def calcular_descuento(total, es_estudiante=False):
    if total >= 50000:
        descuento = 0.10
    elif total >= 30000:
        descuento = 0.05
    else:
        descuento = 0.0
    if es_estudiante:
        descuento += 0.02
    return descuento

# Prueba rápida
p = buscar_producto("P04", catalogo)
print(p)
print("Descuento para $60.000 siendo estudiante:", calcular_descuento(60000, True))
````

````python
%%R
# R
buscar_producto <- function(codigo, catalogo) {
  for (p in catalogo) {
    if (p$codigo == codigo) {
      return(p)
    }
  }
  return(NULL)
}

calcular_descuento <- function(total, es_estudiante = FALSE) {
  if (total >= 50000) {
    descuento <- 0.10
  } else if (total >= 30000) {
    descuento <- 0.05
  } else {
    descuento <- 0.0
  }
  if (es_estudiante) {
    descuento <- descuento + 0.02
  }
  return(descuento)
}

p <- buscar_producto("P04", catalogo)
print(p)
cat("Descuento para $60.000 siendo estudiante:", calcular_descuento(60000, TRUE), "\n")
````

### 8.1 Catálogo y contador de facturas


````python
# PYTHON
CATALOGO = [
    {"codigo": "P01", "nombre": "Cuaderno cuadriculado",  "precio": 3500},
    {"codigo": "P02", "nombre": "Lápiz de mina",           "precio": 800},
    {"codigo": "P03", "nombre": "Borrador",                "precio": 600},
    {"codigo": "P04", "nombre": "Resma de papel carta",    "precio": 15000},
    {"codigo": "P05", "nombre": "Marcador permanente",     "precio": 2200},
    {"codigo": "P06", "nombre": "Tijeras escolares",       "precio": 4200},
]

IVA = 0.19
numero_factura = 1000  # se irá incrementando con cada factura generada
````

### 8.2 Funciones de apoyo

- `mostrar_catalogo()`: imprime el inventario disponible.
- `buscar_producto(codigo)`: retorna el producto o `None` si no existe.
- `calcular_descuento(total, es_estudiante)`: retorna el % de descuento.


````python
# PYTHON
def mostrar_catalogo():
    print("\nCATÁLOGO DISPONIBLE")
    print("-" * 45)
    for p in CATALOGO:
        print(f"{p['codigo']:<5} {p['nombre']:<28} ${p['precio']}")
    print("-" * 45)

def buscar_producto(codigo):
    for p in CATALOGO:
        if p["codigo"].upper() == codigo.upper():
            return p
    return None

def calcular_descuento(total, es_estudiante=False):
    if total >= 50000:
        descuento = 0.10
    elif total >= 30000:
        descuento = 0.05
    else:
        descuento = 0.0
    if es_estudiante:
        descuento += 0.02
    return descuento

mostrar_catalogo()
````

### 8.3 Ciclo principal: agregar productos a la factura

Usamos un ciclo `while` para que el cajero pueda seguir agregando productos **hasta que escriba "fin"**. Este es el corazón interactivo del sistema.

> 💡 Ejecuta la celda y prueba escribiendo códigos como `P01`, `P04`, y al final `fin`.


````python
# PYTHON
def tomar_pedido():
    items_factura = []  # aquí guardamos cada línea de la factura

    print("Escribe el código del producto (ej: P01). Escribe 'fin' para terminar.")
    while True:
        codigo = input("Código de producto: ").strip()

        if codigo.lower() == "fin":
            break

        producto = buscar_producto(codigo)
        if producto is None:
            print("  ⚠️ Producto no encontrado. Intenta de nuevo.")
            continue

        cantidad_str = input(f"  Cantidad de '{producto['nombre']}': ").strip()
        if not cantidad_str.isdigit() or int(cantidad_str) <= 0:
            print("  ⚠️ Cantidad inválida. Intenta de nuevo.")
            continue

        cantidad = int(cantidad_str)
        subtotal = producto["precio"] * cantidad

        items_factura.append({
            "nombre": producto["nombre"],
            "precio": producto["precio"],
            "cantidad": cantidad,
            "subtotal": subtotal
        })
        print(f"  ✅ Agregado: {cantidad} x {producto['nombre']} = ${subtotal}")

    return items_factura

# Descomenta la siguiente línea para probarlo de forma interactiva:
# items = tomar_pedido()
````

### 8.4 Generar la factura completa

Esta función toma la lista de ítems, calcula subtotal, descuento, IVA y total, y **imprime la factura formateada**.


#### 🔎 Chuleta rápida de formato (lo que vas a ver en la factura)

| Código | Significado |
|---|---|
| `:<22` | alinear texto a la izquierda, ancho 22 |
| `:>5` | alinear a la derecha, ancho 5 |
| `:>10.0f` | alinear a la derecha, ancho 10, como decimal con 0 cifras |
| `"=" * 45` | repetir el carácter `=` 45 veces (línea separadora) |

También vas a ver una f-string un poco más compleja:
```python
f"{'Descuento (' + str(int(descuento_pct*100)) + '%):':<32}"
```
Aquí, dentro de las llaves `{ }`, primero **construimos el texto de la etiqueta** (concatenando "Descuento (", el número convertido a texto con `str()`, y "%):" ), y *después* lo alineamos a 32 caracteres con `:<32`. Es una f-string con una expresión adentro, no solo una variable suelta — pero sigue el mismo principio que ya conoces.


````python
# PYTHON
def generar_factura(items_factura, nombre_cliente, es_estudiante=False):
    global numero_factura

    if len(items_factura) == 0:
        print("No hay productos en la factura.")
        return

    subtotal = 0
    for item in items_factura:
        subtotal += item["subtotal"]

    descuento_pct = calcular_descuento(subtotal, es_estudiante)
    valor_descuento = subtotal * descuento_pct
    base_gravable = subtotal - valor_descuento
    valor_iva = base_gravable * IVA
    total = base_gravable + valor_iva

    numero_factura += 1

    print("=" * 45)
    print("      PAPELERÍA EL LÁPIZ FELIZ")
    print(f"      Factura N° {numero_factura}")
    print("=" * 45)
    print(f"Cliente: {nombre_cliente}")
    print("-" * 45)
    print(f"{'Producto':<22}{'Cant':>5}{'Subtotal':>15}")
    for item in items_factura:
        print(f"{item['nombre']:<22}{item['cantidad']:>5}{item['subtotal']:>13}$")
    print("-" * 45)
    print(f"{'Subtotal:':<32}${subtotal:>10.0f}")
    print(f"{'Descuento (' + str(int(descuento_pct*100)) + '%):':<32}-${valor_descuento:>9.0f}")
    print(f"{'IVA (19%):':<32}${valor_iva:>10.0f}")
    print("-" * 45)
    print(f"{'TOTAL A PAGAR:':<32}${total:>10.0f}")
    print("=" * 45)

    return total

# --- EJEMPLO DE PRUEBA (sin necesidad de escribir nada) ---
items_ejemplo = [
    {"nombre": "Cuaderno cuadriculado", "precio": 3500, "cantidad": 5, "subtotal": 17500},
    {"nombre": "Marcador permanente",   "precio": 2200, "cantidad": 3, "subtotal": 6600},
    {"nombre": "Resma de papel carta",  "precio": 15000, "cantidad": 2, "subtotal": 30000},
]
generar_factura(items_ejemplo, "Juan Pérez", es_estudiante=True)
````

````python
# PYTHON
def sistema_facturacion():
    print("BIENVENIDO AL SISTEMA DE FACTURACIÓN")
    print("PAPELERÍA EL LÁPIZ FELIZ")
    mostrar_catalogo()

    nombre_cliente = input("Nombre del cliente: ").strip()
    respuesta_estudiante = input("¿Es estudiante? (si/no): ").strip().lower()
    es_estudiante = respuesta_estudiante == "si"

    items = tomar_pedido()
    generar_factura(items, nombre_cliente, es_estudiante)

# Descomenta para ejecutar el sistema completo de forma interactiva:
# sistema_facturacion()
````

### 8.6 Equivalente en R (cálculo del total, sin `input()` interactivo)

R también puede leer del teclado con `readline()`, pero en Colab funciona mejor cuando el bloque de código no requiere entrada interactiva. Por eso aquí mostramos la **versión de cálculo**: dado un pedido ya armado, R calcula subtotal, descuento, IVA y total — la misma lógica que usamos en Python.


````python
%%R
# R
CATALOGO <- list(
  list(codigo = "P01", nombre = "Cuaderno cuadriculado", precio = 3500),
  list(codigo = "P04", nombre = "Resma de papel carta",  precio = 15000),
  list(codigo = "P05", nombre = "Marcador permanente",   precio = 2200)
)
IVA <- 0.19

calcular_descuento <- function(total, es_estudiante = FALSE) {
  if (total >= 50000) {
    descuento <- 0.10
  } else if (total >= 30000) {
    descuento <- 0.05
  } else {
    descuento <- 0.0
  }
  if (es_estudiante) descuento <- descuento + 0.02
  return(descuento)
}

generar_factura <- function(items, cliente, es_estudiante = FALSE) {
  subtotal <- 0
  cat("PRODUCTO                CANT   SUBTOTAL\n")
  for (item in items) {
    subtotal <- subtotal + item$subtotal
    cat(sprintf("%-24s %-6d $%d\n", item$nombre, item$cantidad, item$subtotal))
  }

  descuento_pct  <- calcular_descuento(subtotal, es_estudiante)
  valor_descuento <- subtotal * descuento_pct
  base_gravable   <- subtotal - valor_descuento
  valor_iva       <- base_gravable * IVA
  total           <- base_gravable + valor_iva

  cat("----------------------------------------\n")
  cat(sprintf("Cliente: %s\n", cliente))
  cat(sprintf("Subtotal: $%.0f\n", subtotal))
  cat(sprintf("Descuento (%.0f%%): -$%.0f\n", descuento_pct * 100, valor_descuento))
  cat(sprintf("IVA (19%%): $%.0f\n", valor_iva))
  cat(sprintf("TOTAL A PAGAR: $%.0f\n", total))
}

items_ejemplo <- list(
  list(nombre = "Cuaderno cuadriculado", cantidad = 5, subtotal = 17500),
  list(nombre = "Marcador permanente",   cantidad = 3, subtotal = 6600),
  list(nombre = "Resma de papel carta",  cantidad = 2, subtotal = 30000)
)

generar_factura(items_ejemplo, "Juan Pérez", es_estudiante = TRUE)
````
