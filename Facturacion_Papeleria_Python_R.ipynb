```python
catalogo = {
    "P-001": {"nombre": "Cuaderno cuadriculado", "precio": 8500,  "iva": 0.19},
    "E-011": {"nombre": "Caja de lapiceros x12", "precio": 15990, "iva": 0.19},
    "L-030": {"nombre": "Libro juvenil",         "precio": 45000, "iva": 0.00},
    "R-050": {"nombre": "Revista educativa",     "precio": 12000, "iva": 0.05},
}
```

se crea un catalogo con nombre y precio

```python
clientes = {
    "C-01": {"nombre": "Doña estel",               "tipo": "detal"},
    "C-02": {"nombre": "Colegio San Andrés",     "tipo": "colegio"},
    "C-03": {"nombre": "Oficina solucionamos S.A.S.", "tipo": "oficina"},
}
```

mis clientes

```python
pedido = [("P-001", 3), ("L-030", 1), ("R-050", 2)]
encabezado = ("F-102", "28/08/2026", "C-02")   # tupla: no se modifica

total = 0
for codigo, cantidad in pedido:
    p = catalogo[codigo]
    subtotal = p["precio"] * cantidad
    iva = subtotal * p["iva"]
    total = total + subtotal + iva
    print(p["nombre"], cantidad, subtotal + iva)

if clientes[encabezado[2]]["tipo"] == "colegio" and total >= 200000:
    total = total * 0.9
print("TOTAL A PAGAR:", total)
```

facturacion

```python
inventario = {"P-001": 0, "E-011": 1, "L-030": 10, "R-050": 20}

codigo = input("Código a consultar: ")
if codigo in inventario and inventario[codigo] > 0:
    print(catalogo[codigo]["nombre"], "→", inventario[codigo], "unidades")
elif codigo in inventario and inventario[codigo] <= 0:
    print ("No hay inventario")
else:
    print("Ese código no existe en la papelería")

for codigo, cantidad in pedido:
    inventario[codigo] = inventario[codigo] - cantidad
```
